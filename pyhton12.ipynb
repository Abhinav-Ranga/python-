# Scope and LEGB Rule
  
A scope is a textual region of a Python program where a namespace is directly accessible.  

The interpreter searches for a name from the inside out, looking in the local, enclosing, global, and finally the built-in scope. If the
interpreter doesn't find the name in any of these locations,then python raises a NameErrorException

# summary
Global namespace exists before you even call anything — it's created the moment the program/module starts running, independent of any function calls.

Local namespace gets created when you call a function — fresh each time, destroyed when that call finishes.

If there's nesting, the innermost function is Local (to itself), and the outer function becomes Enclosing .

global scope is the region of ur global name space
and the varibles u enter inside that region are global variables

```python
# Enclosing scope


# so that means in nested functions all are local  scope except global
# but they can also become enclosed scopes but the inner most can't become
# enclosed

def outer(): # local scope but can become enclosing scope
  def inner(): # local scope
    print("inner function") # runs inside local scope
  inner()# runs inside enclosed scope
  print("outer function")


outer()  # runs inside global scope
print('main program') # runs inside global scope
```

```python
# a is not in inner so by LEGB it will go to enclosed and print 3
def outer():
  a = 3
  def inner():
    print(a)
  inner()
  print("outer function")


outer()
print('main program')
```

# Decorators
A decorator in python is a function that receives another function as input and adds some functionality(decoration) to and it and returns it.

This can happen only because python functions are 1st class citizens.

There are 2 types of decorators available in python

. Built in decorators like @staticmethod, @classmethod, @abstractmethod and @property etc

. User defined decorators that we programmers can create according to our needs

1st class ciitzens are the objects in a programmaing langugage with which we can  perform all operations
ex = we can store them anywhere,give them as input

In python functions are first class citizens

```python
# python are 1st class function



# (1)
# def fun():
#   print('hello')

# a = fun
# a()



# (2)
# def fun():
#   print('hello')

# del fun
# fun()



def modify(func,num):
  return func(num)

def square(num):
    return num ** 2

modify(square,2)
```

```python
# simple example
def my_decorator(func):
  def wrapper():
    print("*****************" )
    func()
    print("*****************")
  return wrapper


def hello():
  print("hello")



def display():
  print('shree harivansh')



a = my_decorator(hello)
a()


b = my_decorator(display)
b()
```

```python
# better syntax

def my_decorator(func):
  def wrapper():
    print("*****************" )
    func()
    print("*****************")
  return wrapper


@my_decorator
def hello():
    print("hello")

hello()
```

```python
# Example


# the thing is we are giving square 2 to timer after returning wrapper now square
# will point  to wrapper not timer so now wrapper can access things from timer but apart from
# this  we have even passed the value 2 in square so that  means  along  with it u are passing
# value 2 to wrapper but in our main function wrapper doesn't take any argument
# so will give us error

import time
def timer(func):
  def wrapper():
    start = time.time()
    func()
    print('time taken by ',func.__name__,'is',time.time()-start,'secs')
  return wrapper


@timer
def square(num):
  time.sleep(1)
  print(num**2)

square()
```

```python

import time
def timer(func):
  def wrapper(* args):
    start = time.time()
    func(*args)
    print('time taken by ',func.__name__,'is',time.time()-start,'secs')
  return wrapper


@timer
def square(num):
  time.sleep(1)
  print(num**2)


def power(a,b):
  print(a**b)


square(1)
power(2,2)
```

```python
# hard example
# but no need to go this deep this is for software engineer

def sanity_check(data_type):
  def outer_wrapper(func):
    def inner_wrapper(*args):
      if type(*args) == data_type:
        func(*args)
      else:
        raise TypeError('ye dataType nahi chalega')
    return   inner_wrapper
  return outer_wrapper




@sanity_check(int)
def square(num):
  print(num **2)


# square(2)

@sanity_check(str)
def greet(name):
  print("hello",name)


greet("shree ji")
```
