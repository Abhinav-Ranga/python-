```python
# when ever u use Exception or any error (ex = zerodivision,filenotfound) iin reality
#  they are classes and u make an object of that class ( e is the object ) then use it


# Exception is main class and
# others are  chlld classes that are inheriting from Exception


class Bank:
  def __init__(self,balance):
    self.balance = balance

  def withdraw(self,amount):
    if amount<0:
      raise Exception("amount can't be negative")
    if self.balance<amount:
      raise Exception("paise nahi hai tere pass: ")
    self.balance = self.balance - amount


obj = Bank(1008)

try:
  obj.withdraw(-1 )
except Exception as e:
  print(e)
else:
  print(obj.balance)

```

```python
# creating custom exceptions

# if u make ur custom exception class u will hava to inherit from Exception class


class MyException(Exception):
  def __init__(self,message):
    print(message)

class Bank:
  def __init__(self,balance):
    self.balance = balance

  def withdraw(self,amount):
    if amount<0:
      raise MyException("amount can't be negative")
    if self.balance<amount:
      raise MyException("paise nahi hai tere pass: ")
    self.balance = self.balance - amount


obj = Bank(1008)

try:
  obj.withdraw(100)
except MyException as e:
  pass
else:
  print(obj.balance)

```
