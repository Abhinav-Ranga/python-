# Using Context Manager (With)
  
It's a good idea to close a file after usage as it will free up the resources  

If we dont close it, garbage collector would close it  

with keyword closes the file as soon as the usage is over

```python
# with read u are writing and closing both together
# with
with open('/content/sample1.txt','w')as f:
  f.write('radha \n')
  f.write('krishna')
```

```python
# seek and tell function
# tell = tells us about our next charcter index no
# seek = seek helps us to decide our starting . without seek the place where we
# leave if we print again then it will continue reading from where we left off but by using
# seek we can actually decide form where we have to start

with open('sample.txt','r') as f:
  print(f.read(10))
  print(f.tell())
  f.seek(4)
  print(f.read(10))

```

# Serialiazation and Deserialization

serialization = process of converting python data types into the JSon format


Deserialization = process of converting json into python data types

 # Json  (Java script object notation) =

 json is the universal data format that is understood by everyone


```python
# serialization using JSON Module
import json
L = [1,2,3,4,5]

with open ('demo.json','w') as f:
  json.dump(L,f)
```

```python
# dictionary

# Json overwrites things nothing to do even though u are using with

d = {
    'abhinav':' ranga',
    'marks':99
}


with open('demo.json','w') as f:
  json.dump(d,f, indent = 4)
```

```python
# deserialization

import json
with open('demo.json','r') as f:
  d = json.load(f)
  print(d)
  print(type(d))
```

# Serializing and deserializing custom object



```python
# if we use this on object it won't work directly we will have to tell in which format we want to write

class Person:
  def __init__(self,fname,lname,age,gender):
    self.fname = fname
    self.lname = lname
    self.age = age
    self.gender = gender
```

```python
person = Person('abhinav','ranga',19,'male')
```

```python
# As a string
import json


def show_obj(person):
  if isinstance(person,Person):
    return "{} {} age>{} gender > {}".format(person.fname,person.lname,person.age,person.gender)


with open('demo.json','w') as f:
  json.dump(person,f,default = show_obj)
```

```python
# As a Dictionary
import json


def show_obj(person):
  if isinstance(person,Person):
    return {
        'fname':person.fname,
        'lname':'person.lname',
        'age':20,
        'gender':person.gender
    }


with open('demo.json','w') as f:
  json.dump(person,f,default = show_obj,indent = 4)
```

```python
# Deserializing
import json

with open('demo.json','r') as f:
  d = json.load(f)
  print(d)
  print(type(d))
```

# Pickling
Pickling is the process whereby a Python object hierarchy is converted into a byte stream, and unpickling is the inverse operation, whereby a byte stream (from a binary file or bytes-like object) is converted back into an object hierarchy.

```python
class Person:
  def __init__(self,name,age):
    self.name = name
    self.age = age


  def display(self):
    return "My name is {} and my age is {} ".format(self.name,self.age)
```

```python
p = Person('abhinav',19)
p.display()
```

```python
# pickle dump
import pickle
with open('pickle.pkl','wb') as f:
  pickle.dump(p,f)
```

```python
# pickle load
with open('pickle.pkl','rb') as f:
  p = pickle.load(f)
  print(p.display())
```

# Pickle vs Json
